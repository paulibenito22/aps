### Introducción
En este trabajo se analiza el comportamiento de distintos circuitos RLC de segundo orden a partir de sus funciones de transferencia. Para ello, los circuitos se estudian en el dominio de Laplace, reemplazando resistores, capacitores e inductores por sus correspondientes impedancias complejas.
A partir de las funciones de transferencia obtenidas se identifican los polos y ceros de cada sistema y se analiza su respuesta en frecuencia, tanto en módulo como en fase. Esto permite clasificar los circuitos según su comportamiento como filtros pasa bajos, pasa altos, pasa banda o rechazo de banda.

Para todos los casos se trabaja con $R=1\,\Omega$, frecuencia natural $\omega_0=1\,\text{rad/s}$ y factor de calidad $Q=\frac{\sqrt2}{2}$. Este último valor corresponde a una respuesta tipo Butterworth de segundo orden, caracterizada por una respuesta en magnitud máximamente plana en la banda pasante.

### 1. Teoría previa
#### Impedancias en Laplace

Dadas las condiciones iniciales nulas, la transformada de Laplace convierte cada derivada en una multiplicación por $s$.

**Para el inductor:**

$$
v(t) = L\,\frac{di}{dt} \;\xrightarrow{\;\mathcal{L}\;}\; V(s) = L\big[sI(s) - \underbrace{i(0)}_{=0}\big] = sL\,I(s) \quad\Rightarrow\quad Z_L = sL
$$

**Para el capacitor:**

$$
i(t) = C\,\frac{dv}{dt} \;\xrightarrow{\;\mathcal{L}\;}\; I(s) = C\big[sV(s) - \underbrace{v(0)}_{=0}\big] = sC\,V(s) \quad\Rightarrow\quad V(s) = \frac{I(s)}{sC} \quad\Rightarrow\quad Z_C = \frac{1}{sC}
$$

**Resistor:** $Z_R = R$.


#### Filtros de segundo orden

Todos los circuitos de esta TS, al ser filtros de segundo orden, tienen el mismo denominador:

$$
D(s) = s^2 + \frac{\omega_0}{Q}\,s + \omega_0^2
$$

- $\omega_0$: frecuencia natural. Es el módulo de los polos (su distancia al origen).
- $Q$: factor de calidad. Mide qué tan cerca del eje $j\omega$ están los polos: a mayor $Q$, más marcado el pico de resonancia.
- $\dfrac{\omega_0}{Q}$: en el pasabanda es el ancho de banda.

Los polos son

$$
s = -\frac{\omega_0}{2Q} \pm j\,\omega_0\sqrt{1 - \frac{1}{4Q^2}}
$$

Con $Q = \frac{\sqrt2}{2}$ queda $\frac{1}{4Q^2} = \frac12$, así que los polos están a $45^\circ$ del eje real.(Es la condición **Butterworth**)

Lo que diferencia un tipo de filtro de otro son los ceros (del numerador), que dependen de donde se toma la salida.

### Datos

$$
R = 1\,\Omega \qquad \omega_0 = 1\ \text{rad/s} \qquad Q = \frac{\sqrt2}{2} \quad\Rightarrow\quad \frac{\omega_0}{Q} = \frac{2}{\sqrt2} = \sqrt2
$$

En los gráficos, $\omega_0 = 1$ rad/s equivale a $f_0 = \dfrac{1}{2\pi} \approx 0{,}159$ Hz.

#### 2.1  Circuito a): pasa bajos
<img src="circuito_a.png" width="400">
A baja frecuencia $C$ es un circuito abierto y $L$ un corto, entonces $V_{out} = V_{in}$. A alta frecuencia $L$ se abre y $C$ se cortocircuita, entonces $V_{out} \to 0$.

#### 2.1.1 Función transferencia:

$$
\begin{aligned}
V_{out} &= i\,Z_C \\
V_{in} &= i\,Z_L + i\,R + i\,Z_C
\end{aligned}
\;\xrightarrow{\;\mathcal{L}\;}\;
\begin{aligned}
V_{out}(s) &= I(s)\left[\tfrac{1}{sC}\right] \\
V_{in}(s) &= I(s)\left[sL + R + \tfrac{1}{sC}\right]
\end{aligned}
$$

$$
T(s) = \frac{\frac{1}{sC}}{sL + R + \frac{1}{sC}} = \frac{\frac{1}{CL}}{s^2 + \underbrace{\frac{R}{L}}_{=\,\omega_0/Q}\,s + \underbrace{\frac{1}{CL}}_{=\,\omega_0^2}}
$$



$$
\begin{cases}
\omega_0 = \dfrac{1}{\sqrt{LC}} = 1 & (1) \\[3mm]
\dfrac{\omega_0}{Q} = \dfrac{R}{L} = \dfrac{2}{\sqrt2} \;\Rightarrow\; L = \dfrac{R\sqrt2}{2} = \dfrac{\sqrt2}{2}\ \text{H}
\end{cases}
$$

$$
\text{en } (1):\quad LC = 1 \;\Rightarrow\; C = \frac{2}{\sqrt2} = \sqrt2\ \text{F}
$$

Reemplazo en $T(s)$:

$$
\boxed{T(s) = \frac{1}{s^2 + \sqrt2\,s + 1}} \qquad \textbf{pasabajos de orden 2}
$$

Como $T(0) = 1 \Rightarrow$ ganancia $= 1$.

####  2.1.2 Polos y ceros

- **Ceros:** no hay ceros finitos (numerador constante).
- **Polos:** 

$$
s^2 + \frac{2}{\sqrt2}\,s + 1 = 0 \;\Rightarrow\; s = -\frac{\sqrt2}{2} \pm j\,\frac{\sqrt2}{2}
$$

Dos polos $\Rightarrow$ pendiente de $-40$ dB/década a alta frecuencia.


#### 2.1.3 Respuesta en frecuencia

Si $s = j\omega$:

$$
T(j\omega) = \frac{1}{-\omega^2 + \sqrt2\,\omega j + 1}
$$

Factorizo $T(s)$ en forma normalizada (útil para leer Bode, porque cada término queda adimensional):

$$
T(s) = \frac{K\omega_0^2}{s^2 + \frac{\omega_0}{Q}\,s + \omega_0^2} = \frac{K}{\frac{s^2}{\omega_0^2} + \frac{s}{Q\,\omega_0} + 1}
$$

$$
|T(j\omega)| = \frac{1}{\sqrt{(1-\omega^2)^2 + 2\omega^2}} = \frac{1}{\sqrt{1+\omega^4}}
$$

$$
\angle T(j\omega) = 0 - \arctan\!\left(\frac{\sqrt2\,\omega}{1-\omega^2}\right)
$$


**En $\omega_0$:**

$$
|T(j\omega_0)| = \frac{1}{\sqrt2} = Q \quad\Rightarrow\quad 20\log(Q) = -3\ \text{dB}
$$

En un pasabajos de segundo orden siempre vale $|T(j\omega_0)| = Q$. Con $Q = \frac{\sqrt2}{2}$, la frecuencia de corte de $-3$ dB coincide justo con $\omega_0$.

####  2.2 Circuito b) 
<img src="circuito_b.png" width="400">
A baja frecuencia $L$ es un corto, entonces $V_{out} \to 0$. A alta frecuencia $C$ cortocircuita a $R$ y $L$ se abre, entonces $V_{out} = V_{in}$.

#### 2.2.1 Función transferencia

$$
V_{out} = i\,Z_L
\qquad\qquad
\text{para } V_{in}:\; \frac{1}{Z_{eq}} = \frac1R + \frac{1}{Z_C}
\;\Rightarrow\;
V_{in} = i\,(Z_{eq} + Z_L)
$$

**Cálculo auxiliar:**

$$
\frac1R + sC = \frac{1 + sCR}{R}
\quad\Rightarrow\quad
\frac{R}{1+sCR} + sL = \frac{R + sL + s^2CRL}{1+sCR}
$$

$$
T(s) = \frac{sL}{\dfrac{s^2CRL + sL + R}{sCR + 1}} = \frac{s^2CRL + sL}{s^2CRL + sL + R} = \frac{s^2 + \frac{s}{CR}}{s^2 + \underbrace{\frac{s}{CR}}_{\omega_0/Q} + \underbrace{\frac{1}{CL}}_{\omega_0^2}} \quad (*)
$$

$$
\begin{cases}
\dfrac{\omega_0}{Q} = \dfrac{1}{CR} = \dfrac{2}{\sqrt2} \;\Rightarrow\; C = \dfrac{\sqrt2}{2}\ \text{F} \\[3mm]
\omega_0 = \dfrac{1}{\sqrt{CL}} = 1 \;\Rightarrow\; L = \dfrac{2}{\sqrt2} = \sqrt2\ \text{H}
\end{cases}
$$

$$
\boxed{T(s) = \underbrace{\frac{s^2}{s^2 + \frac{2}{\sqrt2}\,s + 1}}_{\text{pasaaltos}} + \underbrace{\frac{s\,\omega_0/Q}{s^2 + \frac{2}{\sqrt2}\,s + 1}}_{\text{pasabanda}}}
$$

Para $s = j\omega$:

$$
\boxed{T(j\omega) = \frac{-\omega^2}{(1-\omega^2) + \frac{2}{\sqrt2}\,\omega j} + \frac{\frac{2}{\sqrt2}\,\omega j}{(1-\omega^2) + \frac{2}{\sqrt2}\,\omega j}}
$$

Ganancia de cada término por separado:

$$
\text{para el pasaaltos: } \lim_{s\to\infty} \frac{s^2}{s^2\left(1 + \frac{2}{\sqrt2\,s} + \frac{1}{s^2}\right)} = 1
\qquad\qquad
\text{para el pasabanda: } T(j\omega_0) = 1
$$

$\Rightarrow$ ambos tienen ganancia unitaria en su banda de paso.

Al ser la suma de un pasaaltos y un pasabanda con el mismo denominador. A alta frecuencia domina el pasaaltos ($T \to 1$). Cerca de $\omega_0$ se suman ambos, y eso produce un pico por encima de $0$ dB.

#### 2.2.2 Polos y ceros

- **Polos:** $s^2 + \frac{2}{\sqrt2}\,s + 1 = 0 \;\Rightarrow\; s = -\frac{\sqrt2}{2} \pm j\frac{\sqrt2}{2}$
- 
- **Ceros:** $s^2 + s\,\frac{\omega_0}{Q} = 0 \;\Rightarrow\; s = 0,\quad s = -\frac{\omega_0}{Q} = -\frac{2}{\sqrt2} = -\sqrt2$


#### 2.2.3 Módulo y fase

$$
\boxed{|T(j\omega)| = \frac{\sqrt{\omega^4 + 2\omega^2}}{\sqrt{1+\omega^4}} = \frac{\omega\sqrt{\omega^2+2}}{\sqrt{1+\omega^4}}}
$$

$$
\boxed{\angle T(j\omega) = \arctan\!\left(\frac{\frac{2}{\sqrt2}\,\omega}{-\omega^2}\right) - \arctan\!\left(\frac{\frac{2}{\sqrt2}\,\omega}{1-\omega^2}\right)}
$$

> Los dos $\arctan$ van con cuadrante ($\operatorname{atan2}$): el numerador $-\omega^2 + j\sqrt2\,\omega$ tiene parte real negativa, así que su ángulo está en el segundo cuadrante (entre $90^\circ$ y $180^\circ$). 

#### Cálculo auxiliar: busco el máximo (pico)

Derivo $|T(j\omega)| = \dfrac{\omega\sqrt{\omega^2+2}}{\sqrt{1+\omega^4}}$ con la regla del cociente:

$$
\frac{d}{d\omega}\Big[\omega\sqrt{\omega^2+2}\Big] = \sqrt{\omega^2+2} + \frac{\omega^2}{\sqrt{\omega^2+2}} = \frac{2\omega^2+2}{\sqrt{\omega^2+2}}
\qquad\qquad
\frac{d}{d\omega}\sqrt{1+\omega^4} = \frac{2\omega^3}{\sqrt{1+\omega^4}}
$$

$$
\frac{d|T(j\omega)|}{d\omega}
= \frac{\dfrac{2\omega^2+2}{\sqrt{\omega^2+2}}\,\sqrt{1+\omega^4} - \omega\sqrt{\omega^2+2}\,\dfrac{2\omega^3}{\sqrt{1+\omega^4}}}{1+\omega^4}
= \frac{-2\omega^4 + 2\omega^2 + 2}{(1+\omega^4)\sqrt{\omega^2+2}\,\sqrt{\omega^4+1}} = 0
$$

> **Corrección del paso intermedio:** en las notas aparece $\sqrt{\omega^2+2} + \frac{2\omega^2}{\sqrt{\omega^2+2}}$ (debería ser $\frac{\omega^2}{\sqrt{\omega^2+2}}$) y $\frac{4\omega^3}{\sqrt{1+\omega^4}}$ (debería ser $\frac{2\omega^3}{\sqrt{1+\omega^4}}$, porque la derivada de $\sqrt{u}$ es $\frac{u'}{2\sqrt u}$). El resultado final $-2\omega^4 + 2\omega^2 + 2$ es correcto: con estas dos correcciones, $(2\omega^2+2)(1+\omega^4) - 2\omega^4(\omega^2+2) = -2\omega^4 + 2\omega^2 + 2$.

Llamo $x = \omega^2$:

$$
-2x^2 + 2x + 2 = 0 \;\Rightarrow\; x = \frac{-2 \pm \sqrt{20}}{-4}
\;\Rightarrow\;
\begin{cases}
x = \dfrac{1+\sqrt5}{2} \\[3mm]
x = \dfrac{1-\sqrt5}{2} < 0 \;\rightarrow\; \text{no tiene solución en } \mathbb{R} \text{ para } x = \omega^2
\end{cases}
$$

$$
\omega^2 = \frac{1+\sqrt5}{2} \;\Rightarrow\; \omega \approx 1{,}27 \;(\text{me quedo con el positivo}) \;\Rightarrow\; |T(j\omega)| \text{ tiene un máximo en } \omega = 1{,}27\ \text{rad/s}
$$

#### Valores característicos

$$
|T(0j)| = 0
$$

$$
|T(j\omega_0)| = \sqrt{\frac32} \;\Rightarrow\; \text{en dB: } 20\log\!\left(\sqrt{\tfrac32}\right) = 1{,}76\ \text{dB}
$$

$$
\text{máximo (pico)}: \;|T(1{,}27\,j)| \approx 1{,}27 \;\Rightarrow\; 20\log(1{,}27) \approx 2{,}09\ \text{dB}
$$

$$
\lim_{\omega\to\infty} |T(j\omega)| = \lim_{\omega\to\infty} \frac{\omega^2\sqrt{1 + \frac{2}{\omega^2}}}{\omega^2\sqrt{\frac{1}{\omega^4} + 1}} = 1 \;\Rightarrow\; \text{en dB: } 0\ \text{dB}
$$

#### (*) Factorizo $T(s)$ para la fase

$$
T(s) = \frac{s^2 + \sqrt2\,s}{s^2 + \sqrt2\,s + 1} = \frac{s\,(s + \sqrt2)}{s^2 + \sqrt2\,s + 1} = \frac{s\sqrt2\left(\frac{s}{\sqrt2} + 1\right)}{1 + \sqrt2\,s + s^2}
$$

Cómo aporta cada factor a la fase:

- $s\sqrt2$: cero en el origen. En $j\omega$ es puramente imaginario (sin parte real) $\Rightarrow$ aporta $+90^\circ$ constante.
- $\left(\frac{s}{\sqrt2} + 1\right)$: cero en $-\sqrt2$. Parte real fija y parte imaginaria creciente $\Rightarrow$ va de $0^\circ$ a $90^\circ$; en $\omega = \sqrt2$ vale $45^\circ$.
- $1 + \sqrt2\,s + s^2$: dos polos complejos conjugados $\Rightarrow$ restan de $0^\circ$ a $-180^\circ$.

$$
\angle T(j\omega) = \angle\big(\omega j\sqrt2\big) + \angle\!\left(\frac{\omega j}{\sqrt2} + 1\right) - \angle\Big((1-\omega^2) + \sqrt2\,\omega j\Big)
$$

Valores marcados en el gráfico:

$$
\omega = 1:\quad 90^\circ + 35^\circ - 90^\circ = 35^\circ
\qquad\qquad
\omega = \sqrt2:\quad 90^\circ + 45^\circ - 116{,}6^\circ = 18{,}4^\circ
$$

Para $\omega \to 0$: $90^\circ$. Para $\omega \to \infty$: $90^\circ + 90^\circ - 180^\circ = 0^\circ$.

#### 2.4 Bonus 
<img src="bonus.png" width="400">

#### 2.4.2 Transferencia

$$
V_{in} = i\,(Z_{eq1} + Z_{eq2}) \qquad V_{out} = i\,Z_{eq2}
$$

$$
\frac{1}{Z_{eq1}} = \frac{1}{Z_{L_b}} + \frac{1}{Z_{C_a}}
\qquad\qquad
\frac{1}{Z_{eq2}} = \frac{1}{Z_{L_{1-b}}} + \frac{1}{Z_{C_{1-a}}} + \frac1R
$$


Usando la admitancia $Y = \frac1Z$ (cómoda porque en paralelo las admitancias se suman):

$$
T(s) = \frac{Z_{eq2}}{Z_{eq1} + Z_{eq2}} = \frac{1/Y_2}{1/Y_1 + 1/Y_2} = \frac{1/Y_2}{\frac{Y_2 + Y_1}{Y_1 Y_2}} = \frac{Y_1}{Y_2 + Y_1}
$$

**Cálculos auxiliares:**

$$
\text{para } Y_1:\quad \frac{b}{sL} + sCa
\qquad\qquad
\text{para } Y_2:\quad C(1-a)\,s + \frac{1-b}{sL} + \frac1R
$$

$$
Y_1 + Y_2 = \frac{b}{sL} + sCa + sC(1-a) + \frac{1-b}{sL} + \frac1R = \frac{1}{sL} + sC + \frac1R = \frac{s^2CLR + sL + R}{sLR}
$$

Al sumar, $a$ y $b$ se cancelan: el denominador es el mismo de siempre.

$$
T(s) = \frac{sCa + \frac{b}{sL}}{\dfrac{s^2CLR + sL + R}{sLR}} = \frac{s^2CLRa + bR}{s^2CLR + sL + R} = \frac{s^2a + \frac{b}{CL}}{s^2 + \frac{s}{CR} + \frac{1}{CL}}
$$

$$
\begin{cases}
\dfrac{\omega_0}{Q} = \dfrac{1}{CR} = \dfrac{2}{\sqrt2} \;\Rightarrow\; C = \dfrac{\sqrt2}{2}\ \text{F} \\[3mm]
\omega_0^2 = \dfrac{1}{CL} = 1 \;\Rightarrow\; L = \sqrt2\ \text{H}
\end{cases}
$$

El numerador queda $s^2a + b\,\omega_0^2$ $\Rightarrow$ **filtro notch**.

$$
\boxed{T(s) = \frac{s^2a + b}{s^2 + \sqrt2\,s + 1}} \;\Rightarrow\; \text{filtro notch (rechaza una } \omega\text{)}
$$

#### 2.4.3 Polos y ceros

**Ceros:**

$$
s^2 = -\frac{b}{a} \qquad (b > 0 \text{ y } a > 0 \;\Rightarrow\; -\tfrac{b}{a} < 0)
\;\Rightarrow\; \text{no tiene ceros reales}
\;\Rightarrow\; s = \pm j\sqrt{\frac{b}{a}} = \pm j\,\omega_0\sqrt{\frac{b}{a}} = \pm j\,\omega_r
$$

Los ceros caen **sobre el eje $j\omega$**. Como en los ceros la salida de $T(s)$ es nula $\Rightarrow$ rechaza esas frecuencias ($\omega_r$). Esa es la idea del notch: un cero sobre el eje imaginario anula exactamente una frecuencia.

- si $b > a$: $\omega_r$ queda a la derecha de $\omega_0$ $\Rightarrow$ notch pasabajos.
- si $b < a$: $\omega_r$ queda a la izquierda de $\omega_0$ $\Rightarrow$ notch pasaaltos.
- si $b = a$: $\omega_r$ cae en $\omega_0$.

**Polos:** $s = -\frac{\sqrt2}{2} \pm j\frac{\sqrt2}{2}$ 

#### 2.4.4 Respuesta en frecuencia ($s = j\omega$)

$$
\boxed{T(j\omega) = \frac{-\omega^2a + b}{(1-\omega^2) + \omega j\sqrt2}}
$$

$$
\boxed{|T(j\omega)| = \frac{|b - \omega^2a|}{\sqrt{1+\omega^4}}}
$$

$$
\boxed{\angle T(j\omega) =
\begin{cases}
0^\circ - \arctan\!\left(\dfrac{\sqrt2\,\omega}{1-\omega^2}\right) & \text{si } \omega < \omega_r \\[4mm]
180^\circ - \arctan\!\left(\dfrac{\sqrt2\,\omega}{1-\omega^2}\right) & \text{si } \omega > \omega_r
\end{cases}}
$$

El numerador $b - a\omega^2$ es real: positivo para $\omega < \omega_r$ (ángulo $0^\circ$) y negativo para $\omega > \omega_r$ (ángulo $180^\circ$). Por eso en $\omega = \omega_r$ hay un salto de fase de $0^\circ$ a $180^\circ$.

##### Comportamientos en bajas y altas frecuencias

$$
\lim_{\omega\to0} |T(j\omega)| = b \;\Rightarrow\; \text{los capacitores son abiertos} \;\Rightarrow\; \text{divisor inductivo}
$$

$$
\lim_{\omega\to\infty} |T(j\omega)| = \lim_{\omega\to\infty} \frac{\omega^2\left(a + \frac{b}{\omega^2}\right)}{\omega^2\sqrt{\frac{1}{\omega^4} + 1}} = a \;\Rightarrow\; \text{los inductores son abiertos} \;\Rightarrow\; \text{divisor capacitivo}
$$

##### Casos analizados

Manteniendo:

$$
a=0.5
$$

se analizaron distintos valores de $b$.

| Caso | \(a\) | \(b\) | $\omega_r=\sqrt{b/a}$ | Ganancia en bajas frecuencias | Ganancia en altas frecuencias |
|---|---:|---:|---:|---:|---:|
| \(b<a\) | 0.5 | 0.25 | \(0.707\) rad/s | \(-12.04\) dB | \(-6.02\) dB |
| \(b=a\) | 0.5 | 0.50 | \(1\) rad/s | \(-6.02\) dB | \(-6.02\) dB |
| \(b>a\) | 0.5 | 0.75 | \(1.225\) rad/s | \(-2.50\) dB | \(-6.02\) dB |

Al aumentar $b$ manteniendo fijo $a$, la frecuencia de rechazo se desplaza hacia frecuencias mayores, de acuerdo con:

$$
\omega_r=\sqrt{\frac{b}{a}}
$$
